#### **OptaLive: Quality Assurance & Validation Rules Test Bench**
**Objective:** Test the `OptaLiveValidator` engine on our processed match data (`key_events_3754217.csv`). Evaluate event timestamps, card state consistency, player attributions, substitution completeness, and compute an overall match data-quality score.

#### **Imports & Setup**

```python
import os
import sys
import pandas as pd

sys.path.append(os.path.abspath(".."))

from src.data_loader import OptaLiveLoader
from src.preprocessing import OptaLivePreprocessor
from src.validation import OptaLiveValidator

print("Validation modules imported successfully!")
```

#### **1. Load Processed Events and Lineup Data**
We load the preprocessed key events saved during Phase 2 or run the loader/preprocessor directly for match ID `3754217`.

#### **Load Data & Initialize Validator**

```python
target_match_id = 3754217

# Load via data loader
loader = OptaLiveLoader(raw_data_dir="../data/raw")
df_lineups = loader.load_lineup(target_match_id)
df_events = loader.load_events(target_match_id)

# Run preprocessor to get clean key events
preprocessor = OptaLivePreprocessor(raw_events_df=df_events, lineup_df=df_lineups)
df_key_events = preprocessor.extract_key_events()

# Initialize Validator
validator = OptaLiveValidator(key_events_df=df_key_events, lineup_df=df_lineups)
print(f"Validator initialized with {len(df_key_events)} key events for match {target_match_id}.")
```

#### **2. Execute Automated Validation Checks**
We run `validate_events()` to scan every record for missing timestamps, unlinked players, card anomalies, and incomplete substitutions.

#### **Execute Validation & Display Results**

```python
df_validated = validator.validate_events()

print("Validation Audit Feed (Sample):")
df_validated[['minute', 'event_type', 'team_name', 'player_name', 'validation_status', 'issue_description']].head(10)
```

#### **3. Generate Overall Match Data-Quality Report**
We compute the match health score (0–100%) based on valid, warning, and error frequencies.

#### **Compute Quality Score Metrics**

```python
qa_report = validator.generate_data_quality_report()

print("=" * 40)
print("     MATCH DATA-QUALITY REPORT     ")
print("=" * 40)
print(f"Total Key Events Evaluated : {qa_report['total_events']}")
print(f"Valid Events               : {qa_report['valid_events']}")
print(f"Warnings Flagged           : {qa_report['warnings']}")
print(f"Errors Flagged             : {qa_report['errors']}")
print(f"Overall Match Health Score : {qa_report['quality_score']}%")
print("=" * 40)
```
